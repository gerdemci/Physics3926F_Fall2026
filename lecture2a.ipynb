## Looking Up Documentation with `help()`
`.__doc__` gives you the raw docstring text, but Python's built-in `help()` function is a formatted way to read documentation — for your own functions, Python's built-ins, or an installed library like NumPy.

You can call `help()` two ways:
1. **Inside a notebook, script, or interactive shell** — just call `help(function_name)` like any other function.
2. **From a terminal, with no script at all** — open a terminal, type `python` to start the interpreter, then type `help(function_name)`.

Either way, `help()` works completely offline — it reads the docstring that's already stored in the installed package on your computer. This is a good skill to practice for **midterm**, too!

```python
help(square)
```

### How to Read a `help()` Output
The exact amount of detail depends on how thoroughly the original author wrote the docstring, but most `help()` output follows a similar structure:

- **Signature line** — the function's name and its parameters, e.g. `square(x)`. This tells you what arguments to pass and in what order, and whether any have default values (shown as `arg=default`).
- **Description** — a short explanation of what the function does, taken directly from the docstring.
- **Parameters** — for more thoroughly documented functions (common in libraries like NumPy), each input is listed with its expected type and meaning.
- **Returns** — describes what the function gives back, and its type.
- **Examples** — runnable sample usage you can copy and adapt.
- **Notes / See Also** — extra details or pointers to related functions.

Notice that `help(square)` above is fairly short; that's because we only gave it a one-line docstring. Library functions tend to be much more thoroughly documented, since many people depend on them. Compare:

```python
help(np.zeros)
```

## Readability and Style
Well-documented code is essential for collaboration and future reference. Following best practices for readable code improves maintainability and understanding.

The following examples are from Software Carpentry: The functions `s` and `std_dev` are computationally equivalent (they both calculate the sample standard deviation), but to a human reader, they look very different. 

```python
def s(p):
    a = 0
    for v in p:
        a += v
    m = a / len(p)
    d = 0
    for v in p:
        d += (v - m) * (v - m)
    return np.sqrt(d / (len(p) - 1))
    
print(s([1, 3, 5, 7, 9, 11]))
```

```python
def std_dev(sample):
    '''Calculates the sample standard deviation'''
    sample_sum = 0
    for value in sample:
        sample_sum += value

    sample_mean = sample_sum / len(sample)

    sum_squared_devs = 0
    for value in sample:
        sum_squared_devs += (value - sample_mean) * (value - sample_mean)

    return np.sqrt(sum_squared_devs / (len(sample) - 1))
print(std_dev([1, 3, 5, 7, 9, 11]))
```

You probably found `std_dev` much easier to read and understand than `s`. As this example illustrates, both documentation and a programmer's coding style combine to determine how easy it is for others to read and understand the programmer's code. Choosing meaningful variable names and using blank spaces to break the code into logical "chunks" are helpful techniques for producing readable code. This is useful not only for sharing code with others, but also for the original programmer. If you need to revisit code that you wrote months ago and haven't thought about since then, you will appreciate the value of readable code!

## Dictionaries
A **dictionary** stores values under keys, rather than positions like a list. You access a value with `dict[key]`, check whether a key exists with `in`, and can list all the keys with `.keys()`.

```python
pet_sounds = {
    "dog": "bark",
    "cat": "meow",
    "cow": "moo",
}

print(pet_sounds["cat"])
print("cat" in pet_sounds)
print("lion" in pet_sounds)
print(list(pet_sounds.keys()))
```

## Case-Insensitive String Comparison
String comparisons in Python are case-sensitive by default (`"Yes" != "yes"`). The `.lower()` (or `.upper()`) string method normalizes case before comparing, so you don't need to check every possible capitalization yourself.

```python
answer = "YES"
print(answer == "yes")
print(answer.lower() == "yes")
```
